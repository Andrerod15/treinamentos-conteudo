```python
import csv
import os

arquivo_csv = 'cadastros.csv'
arquivo_corrigido = 'cadastros_corrigido.csv'
cadastros = []

# Se o arquivo existe, lê os cadastros já salvos
if os.path.exists(arquivo_csv):
    with open(arquivo_csv, 'r', newline='', encoding='utf-8') as arquivo:
        reader = csv.DictReader(arquivo, delimiter=';')  # usa o mesmo delimitador que vai usar na gravação
        for linha in reader:
            # Protege a conversão: se idade é número, converte; senão define como 0
            idade_str = linha.get('idade', '').strip()
            if idade_str.isdigit():
                linha['idade'] = int(idade_str)
            else:
                print(f"Atenção: idade inválida ou vazia em linha: {linha}")
                linha['idade'] = 0  # valor padrão se idade estiver ruim
            cadastros.append(linha)

# Coleta novos cadastros
while True:
    cadastro = {}
    cadastro["nome"] = input("Digite o nome: ")
    cadastro["idade"] = int(input("Digite a idade: "))
    cadastro["cidade"] = input("Digite a cidade: ")
    
    cadastros.append(cadastro)
    
    continuar = input("Quer cadastrar outra pessoa? (s/n): ").lower()
    if continuar != 's':
        break

# Salva todos os cadastros (antigos + novos) sobrescrevendo o CSV original em UTF-8
with open(arquivo_csv, 'w', newline='', encoding='utf-8') as arquivo:
    campos = ['nome', 'idade', 'cidade']
    writer = csv.DictWriter(arquivo, fieldnames=campos, delimiter=';')
    
    writer.writeheader()
    for c in cadastros:
        writer.writerow(c)

# Converte o CSV para ANSI para abrir no Excel com acentos certos
with open(arquivo_csv, 'r', newline='', encoding='utf-8') as f_in:
    leitor = list(csv.reader(f_in, delimiter=';'))

with open(arquivo_corrigido, 'w', newline='', encoding='windows-1252') as f_out:
    escritor = csv.writer(f_out, delimiter=';')
    escritor.writerows(leitor)

print(f'Cadastros salvos com sucesso em "{arquivo_csv}" (UTF-8) e "{arquivo_corrigido}" (ANSI para Excel)')
```
