```python
#cadastro de pessoas com salvamento em arquivo csv:

cadastros = []

while True:
    cadastro = {}
    cadastro["nome"] = input("Digite o nome: ")
    cadastro["idade"] = int(input("Digite a idade: "))
    cadastro["cidade"] = input("Digite a cidade: ")
    
    cadastros.append(cadastro)  # adiciona o cadastro na lista
    
    continuar = input("Quer cadastrar outra pessoa? (s/n): ").lower()
    if continuar != 's':
        break


import csv

with open('cadastros.csv', 'w', newline='', encoding='utf-8') as arquivo:
    campos = ['nome', 'idade', 'cidade']
    writer = csv.DictWriter(arquivo, fieldnames=campos)
    
    writer.writeheader()  # escreve cabeçalhos
    for c in cadastros:
        writer.writerow(c)
```

```python
cadastros = []

while True:
    cadastro = {}
    cadastro["nome"] = input("Digite o nome: ")
    cadastro["idade"] = int(input("Digite a idade: "))
    cadastro["cidade"] = input("Digite a cidade: ")
    
    cadastros.append(cadastro)  # adiciona o cadastro na lista
    
    continuar = input("Quer cadastrar outra pessoa? (s/n): ").lower()
    if continuar != 's':
        break


with open('cadastros.csv', 'a', newline='', encoding='utf-8') as arquivo:
    campos = ['nome', 'idade', 'cidade']
    writer = csv.DictWriter(arquivo, fieldnames=campos)

    # Só escreve o cabeçalho se o arquivo está vazio:
    arquivo.seek(0, 2)  # vai pro final do arquivo
    if arquivo.tell() == 0:
        writer.writeheader()
    
    for c in cadastros:
        writer.writerow(c)

```

```python
import csv

def corrigir_caracteres(texto):
    return (
        texto.replace("Ã", "Á").replace("ã", "á")
             .replace("Õ", "Ó").replace("õ", "ó")
             .replace("Ê", "É").replace("ê", "é")
    )

# Lê o arquivo ignorando o DictReader inicialmente
with open('cadastros.csv', 'r', encoding='utf-8') as entrada:
    linhas = list(csv.reader(entrada))  # Lê todas as linhas normalmente

# Processa e corrige os caracteres
dados_corrigidos = []
for i, linha in enumerate(linhas):
    if i == 0:
        dados_corrigidos.append(linha)  # Mantém o cabeçalho
    else:
        dados_corrigidos.append([corrigir_caracteres(celula) for celula in linha])

# Agora escreve de volta com DictWriter (se quiser manter estrutura)
with open('cadastros_corrigido.csv', 'w', newline='', encoding='utf-8') as saida:
    writer = csv.DictWriter(saida, fieldnames=['nome', 'idade', 'cidade'])
    writer.writeheader()
    for linha in dados_corrigidos[1:]:  # Pula o cabeçalho
        writer.writerow({
            'nome': linha[0],
            'idade': linha[1],
            'cidade': linha[2]
        })
```

```python
#correção de caracteres errados:

import csv

# Função para corrigir caracteres errados
def corrigir_caracteres(texto):
    # Substitui caracteres errados por seus equivalentes corretos
    texto_corrigido = (
        texto.replace("Ã", "Á")
             .replace("ã", "á")
             .replace("Õ", "Ó")
             .replace("õ", "ó")
             # Adicione mais substituições conforme necessário
    )
    return texto_corrigido

# Lê o arquivo CSV e corrige os caracteres
with open('cadastros.csv', 'r', encoding='utf-8') as arquivo_entrada:
    leitor = csv.DictReader(arquivo_entrada)
    dados_corrigidos = []
    for linha in leitor:
        # Corrige cada campo do cadastro
        cadastro_corrigido = {
            chave: corrigir_caracteres(valor) for chave, valor in linha.items()
        }
        dados_corrigidos.append(cadastro_corrigido)

# Escreve os dados corrigidos de volta ao arquivo
with open('cadastros.csv', 'w', newline='', encoding='utf-8') as arquivo_saida:
    campos = ['nome', 'idade', 'cidade']
    escritor = csv.DictWriter(arquivo_saida, fieldnames=campos)
    escritor.writeheader()
    escritor.writerows(dados_corrigidos)
```
